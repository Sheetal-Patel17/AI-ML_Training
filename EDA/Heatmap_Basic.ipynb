```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
arr_2D = np.linspace(1,7,12).reshape(4,3)
arr_2D
```

```python
sns.heatmap(arr_2D)
```

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

sns.heatmap(data)

plt.title("Basic Heatmap")
plt.show()
```

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

sns.heatmap(data, annot=True)

plt.title("Heatmap with Annotations")
plt.show()
```

```python
df = pd.DataFrame(
    [[85, 90, 78],
     [88, 76, 95],
     [92, 89, 84]],
    index=["Alice", "Bob", "Charlie"],
    columns=["Math", "Science", "English"]
)

# Draw Heatmap
sns.heatmap(df, annot=True, cmap="coolwarm")

plt.title("Student Marks Heatmap")
plt.show()
```

```python
df = pd.DataFrame(
    [[85, 90, 78],
     [88, 76, 95],
     [92, 89, 84]],
    index=["Alice", "Bob", "Charlie"],
    columns=["Math", "Science", "English"]
)

plt.figure(figsize=(8, 5))
sns.heatmap(df, annot=True, cmap="YlGnBu", fmt=".0f")


plt.title("Student Marks Heatmap")
plt.show()
```

```python
data = np.random.randint(1, 100, (5, 5))

sns.heatmap(data, annot=True, cmap="coolwarm")

plt.title("Heatmap using Coolwarm")
plt.show()
```

```python
data = np.random.rand(6, 6)

sns.heatmap(
    data,
    annot=True,
    cmap="viridis",
    linewidths=1,
    linecolor="black"
)

plt.title("Heatmap with Grid Lines")
plt.show()
```
