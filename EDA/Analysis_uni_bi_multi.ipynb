```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset("tips")
```

```python
df.head()
```

```python
print(df['total_bill'].describe())

plt.figure(figsize=(8, 4))
sns.histplot(df['total_bill'], kde=True, color = "g")
plt.title('Univariate: Ditribution of Total Bill')
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequency')
plt.show()
```

```python
print(df["day"].value_counts())

plt.figure(figsize=(6,4))
sns.countplot(data = df, x = "day", palette="Set2")
plt.title("Univariate: Order Frequency Day by Day")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```

**2. Bivariate Analysis(2 Variable Together)**

```python
#Scatter Plot + Linear Trend Line

plt.figure(figsize=(7, 5))
sns.regplot(data=df, x='total_bill', y='tip',scatter_kws={"alpha":0.9})
plt.title('Bivariate: Total Bill vs Tip Amount')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()

#Pearson Coorelation Coefficient

correlation= df["total_bill"].corr(df["tip"])
print(f"Coorelation between Total Bills and Tips: {correlation:.2f}")
```

**B. Categorical vs. Numerical: Day by Day vs. toal_bill**

```python
#Boxplot t compare distributions across groups
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='day', y='total_bill', palette="Set3")
plt.title('Bivariate: Total Bill by Day of Week')
plt.xlabel('Day')
plt.ylabel('Total Bill ($)')
plt.show()
```

**3. Multivariate Analysis**

```python
#Scatter plot hue mapping
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='total_bill', y='tip', hue='smoker', style ='smoker',s =70)

plt.title('Multivariate: Total Bill vs Tip by Day and Smoker Status')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.legend(title='Smoker')
plt.show()
```

```python
#Compute correlation matrix for numeric columns

numeric_df = df.select_dtypes(include=["float64","int64"])
correlation_matrix = numeric_df.corr()

#Heatmap visualization
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Multivariate: Correlation Heatmap')
plt.show()
```
