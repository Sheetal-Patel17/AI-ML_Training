```python
classes = ['Python', 'ML', 'DS', 'Cloud Computing', 'DBMS']
class1_student = [3,20,30,15,50]
explode = [0, 0, 0.2, 0, 0]
colors = ['c','b','r','y','g']
textprops = {'fontsize':15}

plt.pie(class1_student, labels = classes, explode = explode, colors = colors, autopct = '% 0.1f %%', shadow = True, radius = 1.4, startangle = 200)
plt.show()
```

```python
classes = ['Python', 'ML', 'DS', 'Cloud Computing', 'DBMS']
class2_student = [20,12,40,21,10]
explode = [0, 0, 0, 0.1, 0]
plt.pie(class1_student, labels = classes, explode = explode)
plt.show
```

```python
wedgeprops = {'linewidth' :0.5, 'width' : 1, 'edgecolor' :'k'}
plt.pie(class1_student,
        labels = classes,
        explode = explode,
        colors= colors,
        autopct= '% 0.2f %%',
        shadow = True,
        radius = 1,
        startangle = 90,
        counterclock = True,
        labeldistance = 1.5,
        wedgeprops = wedgeprops,
        textprops = textprops,
        center = (0.5,45),
        frame = True,
        rotatelabels = True)
plt.show()
```
