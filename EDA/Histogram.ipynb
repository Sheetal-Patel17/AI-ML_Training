```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
import random
```

```python
ml_student_age = np.random.randint(18,28,(100))
py_student_age = np.random.randint(18,28,(100))
```

```python
ml_student_age
```

```python
py_student_age
```

```python
plt.hist(ml_student_age)
plt.title('ML Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.show()
```

```python
plt.hist(py_student_age,)
plt.title('PY Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.show()
```

```python
plt.hist(ml_student_age,rwidth=0.8)
plt.title('ML Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.show()
```

```python
#rwidth
ml_student_age = np.random.randint(18,28,(100))
py_student_age = np.random.randint(18,28,(100))

bins= [15,20,25,30,35,40,45]

plt.hist(ml_student_age,bins)
plt.title('ML Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.show()
```

```python
#histtype
ml_student_age = np.random.randint(18,28,(100))
py_student_age = np.random.randint(18,28,(100))

bins= [15,20,25,30,35,40,45]

plt.hist(ml_student_age,bins, rwidth =0.9, histtype= 'step')

plt.title('ML Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.show()
```

```python
#histtype
ml_student_age = np.random.randint(18,28,(100))
py_student_age = np.random.randint(18,28,(100))

bins= [15,20,25,30,35,40,45]

plt.hist(ml_student_age,bins, rwidth =0.9, histtype= 'barstacked')
plt.title('ML Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.show()
```

```python
#align
ml_student_age = np.random.randint(18,50,(100))
py_student_age = np.random.randint(18,45,(100))

bins= [15,20,25,30,35,40]

plt.hist(ml_student_age,bins, rwidth =0.9, histtype= 'barstacked', align = 'right')
plt.title('ML Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.show()
```

```python
#orientation
ml_student_age = np.random.randint(18,50,(100))
py_student_age = np.random.randint(18,45,(100))

bins= [15,20,25,30,35,40]

plt.hist(ml_student_age,bins, rwidth =0.9, histtype= 'barstacked', align = 'right', orientation = 'horizontal', color = 'cyan')
plt.title('ML Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.show()
```

```python
plt.hist(ml_student_age,bins, rwidth =0.9, histtype= 'barstacked', color = 'r', label = 'ML')
```

```python
bins=[15,20,25,30,35,40,45,50,55,60,65,70,75]

plt.hist(ml_student_age, bins, rwidth =0.8, histtype= 'barstacked', color = 'r', label = 'ML')
plt.hist(py_student_age, bins, rwidth =0.8, histtype= 'barstacked', color = 'g', label = 'PY')

plt.title('ML & PY Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.legend()
plt.show()
```

```python
bins=[15,20,25,30,35,40,45,50,55,60,65,70,75]
plt.hist(ml_student_age, bins, rwidth =0.8, histtype= 'barstacked', orientation = 'horizontal', color = 'r', label = 'ML')
plt.hist(py_student_age, bins, rwidth =0.8, histtype= 'barstacked', orientation = 'horizontal', color = 'g', label = 'PY')

plt.title('ML & PY Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.legend()
plt.show()
```

```python
bins=[15,20,25,30,35,40]
plt.hist(ml_student_age, bins, rwidth =0.8, histtype= 'barstacked', orientation = 'vertical', color = 'r', label = 'ML')
plt.hist(py_student_age, bins, rwidth =0.8, histtype= 'barstacked', orientation = 'horizontal', color = 'm', label = 'PY')

plt.title('ML & PY Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.legend()
plt.show()
```

```python
bins=[15,20,25,30,35,40]
plt.hist([ml_student_age, py_student_age] ,bins, rwidth =0.8, histtype= 'bar',color = ['r','m'], label = ['ML','PY'])

plt.title('ML & PY Student Age' )
plt.xlabel('Student Age')
plt.ylabel('Student Count')
plt.grid(color = 'pink',linestyle =':', linewidth = 0.5)
plt.legend()
plt.show()
```
