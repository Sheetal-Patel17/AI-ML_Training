- Heatmap is a graphical representation of 2D data.Each data value represents in a matrix and it has a special colour.


A heatmap is used when you want to visualize values using colors. The darker or brighter the color, the higher or lower the value. It helps you quickly identify patterns, correlations, or areas of high and low intensity.


## Used in:
- Exploratory Data Analysis (EDA)
- Feature Selection
- Detecting multicollinearity


| Heatmap Use        | Why It Is Used                         |
| ------------------ | -------------------------------------- |
| Correlation Matrix | Find relationships between features    |
| Missing Values     | Identify null values visually          |
| Confusion Matrix   | Evaluate classification models         |
| Sales Dashboard    | Compare sales across products and time |
| Student Marks      | Compare performance across subjects    |
| Website Traffic    | Identify peak activity times           |

### Step 1: What is a Heatmap
- Imagine you have marks of students

| Student | Math |
| ------- | ---- |
| A       | 95   |
| B       | 70   |
| C       | 45   |
| D       | 88   |


- 95  ██████████  (Dark Green)
- 88  ████████
- 70  █████
- 45  ██


- A heatmap simply converts numbers into colors.

1. Large number → Dark color
2. Small number → Light color

```python
import pandas as pd

df = pd.DataFrame({
    "Math": [95, 80, 75, 60, 45],
    "Science": [90, 78, 70, 65, 50],
    "English": [88, 82, 76, 55, 48]
})

print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.heatmap(df)

plt.show()
```

```python
sns.heatmap(df,
            annot=True)

plt.show()
```

```python
# change colour theme

sns.heatmap(df,
            annot=True,
            cmap="coolwarm")
```

```python
# adding grid lines

sns.heatmap(
    df,
    annot=True,
    linewidths=1,
    linecolor="black"
)
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(df,
            annot=True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(df,
            annot=True,
            cmap="YlGnBu")

plt.title("Student Marks Heatmap")

plt.show()
```

## What Does Each Parameter Mean?


| Parameter           | Purpose                  |
| ------------------- | ------------------------ |
| `df`                | Dataset to visualize     |
| `annot=True`        | Show values inside cells |
| `cmap="YlGnBu"`     | Color theme              |
| `linewidths=1`      | Cell border thickness    |
| `linecolor="black"` | Border color             |


# How to Read Correlation Values

| Correlation     | Meaning                        |
| --------------- | ------------------------------ |
| **1.0**         | Perfect positive relationship  |
| **0.8 to 0.99** | Strong positive relationship   |
| **0.5 to 0.79** | Moderate positive relationship |
| **0.0**         | No relationship                |
| **-0.5**        | Moderate negative relationship |
| **-1.0**        | Perfect negative relationship  |
