```python
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
tips_data_f = sns.load_dataset('tips')
```

```python
tips_data_f
```

```python
sns.barplot()
```

```python
sns.barplot(x = tips_data_f.total_bill)
```

```python
sns.barplot(x = tips_data_f['total_bill'], color = 'r')
```

```python
sns.barplot(x = tips_data_f['day'], y = tips_data_f['total_bill'])
```

```python
sns.barplot(x = tips_data_f.day, y = tips_data_f.total_bill)
```

```python
sns.barplot(x = tips_data_f.day, y = tips_data_f.total_bill, hue = tips_data_f.sex)
```

```python
sns.barplot(x = 'day', y = 'total_bill', hue = 'sex', data = tips_data_f)
```

```python
order = ['Sun', 'Thur' , 'Fri', 'Sat']

sns.barplot(x = 'day', y = 'total_bill', hue = 'sex', data = tips_data_f,
           order = order)
```

```python
# male female to female male
```

```python
order = ['Sun', 'Thur' , 'Fri', 'Sat']
hue_order = ['Female', 'Male']


sns.barplot(x = 'day', y = 'total_bill', hue = 'sex', data = tips_data_f,
           order = order, hue_order =  hue_order)
```
