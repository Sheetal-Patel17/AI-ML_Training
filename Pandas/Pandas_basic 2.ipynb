#Data Structures: Series and DataFrame AND PANEL

Pandas provides two core data structures:

- Series: A one-dimensional array WITH HEADER(labeled) AND INDEX (like a column in a spreadsheet).
- DataFrame: A two-dimensional table with rows and columns.
(Series + Series = DataFrame)


- Series
> A Series is a one-dimensional labeled array capable of holding any data type.


> NUMPY - Numpy array is used for the implementation of pandas data objects.

```python
# Create a Series
s = pd.Series([1, 2, 3, 4], index=['a', 'b', 'c', 'd'])
print(s)

s3 = pd.Series([1,2,3,4], index = ['a','b','c','d'],dtype = float)
print(s3)
print('\n')


s2 = pd.Series([1,2,3,4], index = ['z','d','b','a'], dtype = float, name = 'data values')
print(s2)
```

#DataFrame
> Pandas DataFrame is two-dimenstional, size-mutable, potentially heterogeneous tabular data structure with labeled axes (rows and columns)

```python
emt_df = pd.DataFrame()
print(emt_df)
```

```python
lst = ['a','b','c']
print(lst)
```

```python
df = pd.DataFrame(lst)
print(df)
df
```

```python
lst_of_lst = [[1,2,3],[2,3,4],[4,5,6]]
print(lst_of_lst)
```

```python
df1 = pd.DataFrame(lst_of_lst)
print(df1)
```

```python
dict1 = {'ID' : [11,22,33,44], 'SN' : [1,2,3,4]}
dict1
```

```python
df2 = pd.DataFrame(dict1)
df2
```

```python
lst_dict = [{'a' : 1, 'b':2}, {'a':3,'b':4}]
df3 = pd.DataFrame(lst_dict)
df3
```

```python
lst_dict = [{'a' : 1, 'b':2}, {'a':3,'b':4, 'c':5}]
df4 = pd.DataFrame(lst_dict)
df4
```

```python
dict_sr = {'ID' : pd.Series([1,2,3]), 'SN' :pd.Series([111,222,333])}
df5 = pd.DataFrame(dict_sr)
df5
```

```python

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'Salary': [50000, 60000, 70000]
}

df = pd.DataFrame(data)
print(df)
```
