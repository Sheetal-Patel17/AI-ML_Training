```python
import numpy as np

arr_1 = np.array([10, 20, 30, 40, 50])
arr_2 = np.array([11])

print("Addition:", arr_1 + arr_2)

print("Subtraction:", arr_1 - arr_2)

print("Multiplication:", arr_1 * arr_2)

print("Division:", arr_1 / arr_2)

print("Floor Division:", arr_1 // arr_2)

print("Modulus:", arr_1 % arr_2)

print("Power:", arr_1 ** 2)

print("Data Type of arr_1:", arr_1.dtype)
print("Data Type of arr_2:", arr_2.dtype)
```

```python
import numpy as np

sheetal = np.array([
    [1, 3, 7],
    [4, 5, 6],
    [9, 10, 15]
])

sheetal_2 = np.array([55])

sum_arr = sheetal + sheetal_2

print("Addition:")
print(sum_arr)

print("Subtraction:")
print(sheetal - sheetal_2)

print("Multiplication:")
print(sheetal * sheetal_2)

print("Division:")
print(sheetal / sheetal_2)

print("Floor Division:")
print(sheetal // sheetal_2)

print("Modulus:")
print(sheetal % sheetal_2)

print("Power:")
print(sheetal ** 2)
```

```python
import matplotlib.pyplot as plt
import numpy as np

arr = np.array([
    [1, 3, 4],
    [4, 7, 6],
    [7, 9, 5]
])

plt.plot(arr[0], marker='o', label='Row 1')
plt.plot(arr[1], marker='o', label='Row 2')
plt.plot(arr[2], marker='o', label='Row 3')

plt.title("2D Array Graph")
plt.xlabel("Column")
plt.ylabel("Value")
plt.legend()

plt.show()
```

```python
import numpy as np

sheetal = np.array([10, -5, 20, -15, 30, -25, 40, -35, 50, -45, 60, -55, 45, -1])

positive = sheetal[sheetal > 0]
negative = sheetal[sheetal < 0]

print("Original Array:", sheetal)
print("Positive Values:", positive)
print("Negative Values:", negative)
```
