```python
import matplotlib.pyplot as plt
from matplotlib import style

# Data
classes = ['Python', 'ML', 'DS', 'Cloud Computing', 'DBMS']
class1_student = [3, 20, 30, 15, 50]

# Apply style
style.use('ggplot')

# Figure size
plt.figure(figsize=(15,5))

# Horizontal Bar Graph
plt.barh(classes,
         class1_student,
         align='edge',
         color='m',
         edgecolor='r',
         linewidth=5,
         alpha=0.5,
         linestyle='--',
         label='Class 1')

# Title and Labels
plt.title('Number of Students', fontsize=15)
plt.xlabel('Number of Students', fontsize=15)
plt.ylabel('Classes', fontsize=15)

# Legend
plt.legend()

# Show graph
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

classes = ['Python', 'ML', 'DS', 'Cloud Computing', 'DBMS']
class1_student = [3, 20, 30, 15, 50]
class2_student = [20, 12, 40, 21, 10]

classes_index = np.arange(len(classes))
width = 0.35

plt.figure(figsize=(15, 10))

# Class 1 bars
plt.bar(classes_index,
        class1_student,
        width=width,
        color='b',
        label='Class 1 Students')

# Class 2 bars
plt.bar(classes_index + width,
        class2_student,
        width=width,
        color='m',
        label='Class 2 Students')

plt.xticks(classes_index + width/2, classes)

plt.title('Bar Chart of MU Class', fontsize=15)
plt.xlabel('Classes', fontsize=10)
plt.ylabel('Number of Students', fontsize=10)

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
plt.figure(figsize=(15,10))
class1_student = [3,20,30,15,50]
class2_student = [20,12,40,21,10]
class3_student = [10,40,5,23,17]

classes_index = np.arange(len(classes))
width = 0.2

plt.bar(classes_index, class1_student, width = width, color = 'b')
plt.bar(classes_index + width, class2_student, width = width, color = 'm')
plt.bar(classes_index + width + width, class3_student, width = width, color = 'g')

plt.xticks(classes_index + width, classes)

plt.title('Bar chart of MU class', fontsize = 20)
plt.xlabel('Classes', fontsize = 15)
plt.ylabel('Number of students', fontsize = 30)
```
