Removing Duplicates

```python
data = {
    'Name': ['Alice','Bob','Alice','Alice','Charlie'],
    'Score':[25, 30, 25, 26, 35],
    'City': ['New York', 'London', 'New York', 'Tokyo', 'Sydney']
}

dfr = pd.DataFrame(data)
print("Original DataFrame:")
print(dfr)
print("\n" + "="*40 + "\n")
```

```python
#2. Remove duplicates rows

dfr_no_duplicates = dfr.drop_duplicates()
print("DataFrame after removing duplicates:")
print(dfr_no_duplicates)
print("\n" + "="*40 + "\n")
```

```python
#Remove duplicates based on a specific column

dfr_no_duplicates_name = dfr.drop_duplicates(subset=['Name'])
print("DataFrame after removing duplicates based on 'Name' column:")
print(dfr_no_duplicates_name)
print("\n" + "="*40 + "\n")
```

```python
#Remove duplicates and kepp the LAST occurrence

df_keep_last = dfr.drop_duplicates(keep='last')
print("DataFrame after removing duplicates and keeping the last occurrence:")
print(df_keep_last)
print("\n" + "="*40 + "\n")
```

```python
#Drop all occurrenced of duplicates completely

dfr_drop_all = dfr.drop_duplicates(keep=False)
print("DataFrame after dropping all occurrences of duplicates:")
print(dfr_drop_all)
print("\n" + "="*40 + "\n")
```

```python
#Modify the original DataFrame in-place and rese the index

dfr.drop_duplicates(inplace=True, ignore_index=True)
print("Modified DataFrame after removing duplicates and resetting the index:")
print(dfr)
print("\n" + "="*40 + "\n")
```
