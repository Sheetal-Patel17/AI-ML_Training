## pd.get_dummies() vs OneHotEncoder

| Feature                         | `pd.get_dummies()` | `OneHotEncoder`                       |
| ------------------------------- | ------------------ | ------------------------------------- |
| Part of scikit-learn pipeline   | ❌                  | ✅                                     |
| Fit on training, transform test | ❌                  | ✅                                     |
| Handles unknown categories      | ❌                  | ✅ (`handle_unknown`)                  |
| Groups rare categories          | ❌                  | ✅ (`min_frequency`, `max_categories`) |
| Sparse output support           | Limited            | ✅                                     |
| Easy for EDA                    | ✅                  | ❌                                     |
| Recommended for ML pipelines    | ❌                  | ✅                                     |


## 1. categories
Purpose

Specifies which categories should be encoded.

Default

categories='auto'

This means the encoder automatically detects unique values.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

```

```python
df = pd.DataFrame({
    'Color':['Red','Blue','Green']
})

encoder = OneHotEncoder()
df = encoder.fit(df)
df
```

```python
df
```

##### Manual Categories

Now it will always expect these categories.

Useful when

Training and testing datasets differ.
You want fixed columns.

Example :- 

**Training**

Blue
Red

**Testing**

Green

>Without predefined categories, Green could cause issues depending on handle_unknown.

```python
encoder = OneHotEncoder(
    categories=[['Blue','Green','Red']]
)
```

## 2. drop
Purpose

Drops one category from each feature.

Default : drop=None

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    sparse_output=False,
    dtype=int
)

encoder.fit_transform([['Male'], ['Female'], ['Male'], ['Female']])
```

```python
encoder = OneHotEncoder(
    drop='first',
    sparse_output=False,
    dtype=int
)
```
