| Parameter    | Default  | Purpose                               | When to Use                                           |
| ------------ | -------- | ------------------------------------- | ----------------------------------------------------- |
| `data`       | Required | Input DataFrame/Series                | Always                                                |
| `prefix`     | `None`   | Custom prefix for dummy columns       | Avoid column name conflicts                           |
| `prefix_sep` | `'_'`    | Separator between prefix and category | Customize column names                                |
| `dummy_na`   | `False`  | Create a separate column for NaN      | Missing values are meaningful                         |
| `columns`    | `None`   | Encode only specified columns         | Selective encoding                                    |
| `sparse`     | `False`  | Use sparse storage                    | Very large datasets with many zeros                   |
| `drop_first` | `False`  | Drop the first dummy column           | Linear/Logistic Regression to avoid multicollinearity |
| `dtype`      | `None`   | Output data type                      | Use `int` or `uint8` for ML datasets                  |


```python
import pandas as pd
```

```python
pd.get_dummies(
    data,
    prefix=None,
    prefix_sep='_',
    dummy_na=False,
    columns=None,
    sparse=False,
    drop_first=False,
    dtype=None
  )
```

## 3. prefix_sep

```python
pd.get_dummies(
    df,
    prefix='Color',
    prefix_sep='-',
    dtype=int
)
```

## 4. dummy_na

Purpose

Creates a separate column for missing (NaN) values.

```python
df1 = pd.DataFrame({
    'Color':['Red','Blue',None]
})
```

```python
pd.get_dummies(df1, dtype=int)
```

## 5. columns

Purpose

Specifies which columns should be encoded.

Default
columns=None

```python
df = pd.DataFrame({
    'Age':[20,21],
    'Color':['Red','Blue'],
    'Gender':['M','F']
})
```

```python
df
```

```python
pd.get_dummies(
    df,
    columns=['Color'],
    dtype=int
)
```

```python
pd.get_dummies(
    df,
    columns=['Color','Gender'],
    dtype=int
)
```

## 6. sparse
Purpose

Stores dummy variables in a sparse format to save memory.

Default : False

Useful when

- Millions of rows
- Thousands of categories
- Mostly 0 values

```python
pd.get_dummies(
    df,
    sparse=True
)
```

## 7. drop_first
Purpose

Drops the first dummy column.

Default

```python
df = pd.DataFrame({
    'Gender':['Male','Female','Male']
})
```

```python
pd.get_dummies(df, dtype=int)
```

```python
pd.get_dummies(
    df,
    drop_first=True,
    dtype=int
)
```
