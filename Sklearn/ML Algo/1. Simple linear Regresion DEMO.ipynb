**Step 1: Import Libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt

```

**Step 2: Create Dataset**

```python
data = {
    "Area": [1000, 1500, 2000, 2200, 2400, 2500],
    "Price": [30, 45, 60, 64, 66,69]
}

df = pd.DataFrame(data)

print(df)
```

**Step 3: Separate Independent and Dependent Variables**

```python
X = df[["Area"]]     # Independent Variable
y = df["Price"]      # Dependent Variable

# X must be in 2D
```

**Step 4: Create Model**

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

**Step 5: Train Model**

```python
model.fit(X, y)

# Slope (m)
# Intercept (c)
```

**Step 6: Print Equation**

```python
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)


# Price = 0.02 × Area + 5.44
```

**Step 7: Predict New Price**

```python
prediction = model.predict([[1800]])

print("Predicted Price:", prediction[0], "Lakhs")
```

**Step 8: Predict Multiple Houses**

```python
new_area = [[1200], [1700], [2500]]

predictions = model.predict(new_area)

print(predictions)
```

**Step 9: Plot Regression Line**

```python
plt.figure(figsize=(7,5))

plt.scatter(X, y, color="blue", label="Actual Data")

plt.plot(X, model.predict(X),
         color="red",
         linewidth=2,
         label="Regression Line")

plt.xlabel("Area (sq ft)")
plt.ylabel("Price (Lakhs)")
plt.title("Simple Linear Regression")
plt.legend()

plt.show()
```

**Step 10: Evaluate Model**

- MAE: Average prediction error in the same units as the target.
- MSE: Penalizes larger errors more heavily.
- RMSE: Typical prediction error.
- R²: How much of the target's variation is explained by the model (closer to 1 is better).


#### R² Score (Coefficient of Determination)
- 1     → Perfect prediction
- 0.9   → Very Good
- 0.8   → Good
- 0.5   → Average
- 0     → Model is no better than predicting the mean
- <0    → Very Bad Model

```python
from sklearn.metrics import r2_score

pred = model.predict(X)

print("R² Score:", r2_score(y, pred))
```
