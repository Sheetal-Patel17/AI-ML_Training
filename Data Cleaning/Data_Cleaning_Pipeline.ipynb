```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
#load data set
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")
print('shape of train dataset: ',train.shape)
print('shape of test dataset; ',test.shape)
```

```python
X_train = train.drop(columns = "SalePrice", axis = 1)
y_train = train['SalePrice']
X_test = test.copy()
print('Shape of X_train : ',X_train.shape)
print('Shape of y_train : ',y_train.shape)
print('Shape of X_test : ',X_test.shape)
```

```python
isnull_sum = X_train.isnull().sum()
isnull_sum
```

```python
X_train.head(2)
y_train.head(2)
X_test.head(2)
```

Numerical value accessing

```python
num_vars = X_train.select_dtypes(include = ['int64', 'float64']).columns
num_vars
```

Missing Value Impution

```python
num_var_miss = [var for var in num_vars if isnull_sum[var] > 0]
num_var_miss
```

```python
cat_vars = X_train.select_dtypes(include = ['object']).columns
cat_vars
```

```python
cat_var_miss = [var for var in cat_vars if isnull_sum[var] > 0]
cat_var_miss
```

```python
num_vars_mean = ['LotFrontage']
num_vars_median = ['MasVnrArea', 'GarageYrBlt']
cat_vars_mode = ['Alley',
 'MasVnrType',
 'BsmtQual',
 'BsmtCond',
 'BsmtExposure',
 'BsmtFinType1',
 'BsmtFinType2',
 'Electrical',
 'FireplaceQu']

#constant value
cat_vars_constant_missing = ['GarageType',
 'GarageFinish',
 'GarageQual',
 'GarageCond',
 'PoolQC',
 'Fence',
 'MiscFeature']
```

Pipeline: Pipeline(Shift + Tab)

```python
num_var_mean_imputer = Pipeline(steps = [('imputer', SimpleImputer(strategy = 'mean'))])
num_var_median_imputer = Pipeline(steps = [('imputer', SimpleImputer(strategy = 'median'))])
cat_vars_mode_imputer = Pipeline(steps = [('imputer', SimpleImputer(strategy = 'most_frequent'))])
cat_vars_constant_missing_imputer = Pipeline(steps = [('imputer', SimpleImputer(strategy = 'constant', fill_value = 'missing'))])
```

```python
num_var_mean_imputer.fit(X_train[num_vars_mean])
```

```python
num_var_mean_imputer
```

```python
num_var_median_imputer
```

```python
cat_vars_mode_imputer
```

```python
cat_vars_constant_missing_imputer
```

ColumnTransformer

```python
preprocessor = ColumnTransformer(transformers = [('mean_imputer', num_var_mean_imputer, num_vars_mean),
                                                 ('median_imputer', num_var_median_imputer, num_vars_median),
                                                 ('mode_imputer', cat_vars_mode_imputer, cat_vars_mode),
                                                 ('constant_imputer', cat_vars_constant_missing_imputer, cat_vars_constant_missing)])
```

Appling stratergy of dataset

```python
preprocessor.fit(X_train)
```

Which stratergy applied how to check

```python
preprocessor.named_transformers_['mean_imputer'].named_steps['imputer'].statistics_
```

```python
X_train_clean = preprocessor.transform(X_train)
X_train_clean
```
