```python
missing_val_perc = df.isnull().sum()/df.shape[0] * 100
missing_val_perc
```

```python
miss_val_clm_gre_20 = missing_val_perc[missing_val_perc > 20].keys()
miss_val_clm_gre_20
```

```python
df2_drop_clm = df.drop(columns = miss_val_clm_gre_20)
df2_drop_clm.shape
```

```python
df3_numeric = df2_drop_clm.select_dtypes(include = ['int64', 'float64'])
df3_numeric.head()
```

```python
plt.figure(figsize = (20,20))
sns.heatmap(df3_numeric.isnull())
```

```python
df3_numeric[df3_numeric.isnull().any(axis = 1)]
```

```python
df3_numeric.isnull().sum()
```

```python
missing_num_var = [var for var in df3_numeric.columns if df3_numeric[var].isnull().sum() > 0]
missing_num_var
```

```python
plt.figure(figsize = (20,20))
sns.set()
for i , var in enumerate(missing_num_var):
    plt.subplot(2,2,i+1)
    sns.distplot(df3_numeric[var], bins = 20, kde_kws = {'linewidth' : 2, 'color' : '#DC143C'})
    
```

```python
df4_num_mean = df3_numeric.fillna(df3_numeric.mean())
df4_num_mean.isnull().sum().sum()
```

```python
plt.figure(figsize = (20,20))
sns.set()
for i , var in enumerate(missing_num_var):
    plt.subplot(2,2,i+1)
    sns.distplot(df3_numeric[var], bins = 20, kde_kws = {'linewidth' : 2, 'color' : '#DC143C'}, label = 'Original')
    sns.distplot(df4_num_mean[var],bins = 20, kde_kws = {'linewidth' : 3, 'color' : 'g'},label = 'mean')
    plt.legend()
```

```python
df5_num_median = df3_numeric.fillna(df3_numeric.median())
df5_num_median.isnull().sum().sum()
```

```python
plt.figure(figsize = (20,20))
sns.set()
for i , var in enumerate(missing_num_var):
    plt.subplot(2,2,i+1)
    sns.distplot(df3_numeric[var], bins = 20, kde_kws = {'linewidth' : 2, 'color' : '#DC143C'}, label = 'Original')
    sns.distplot(df4_num_mean[var],bins = 20, kde_kws = {'linewidth' : 3, 'color' : 'g'},label = 'mean')
    sns.distplot(df5_num_median[var],bins = 20, kde_kws = {'linewidth' : 3, 'color' : 'k'},label = 'median')
    plt.legend()
```

```python
for i, var in enumerate(missing_num_var):
    plt.figure(figsize = (5,9))
    plt.subplot(3,1,1)
    sns.boxplot(df[var])
    plt.subplot(3,1,2)
    sns.boxplot(df4_num_mean[var])
    plt.subplot(3,1,3)
    sns.boxplot(df5_num_median[var])
```

```python
for i, var in enumerate(missing_num_var):
    plt.figure(figsize=(20,11))
    plt.subplot(3, 1, 1)
    sns.boxplot(x=df[var], orient='h')
    plt.subplot(3, 1, 2)
    sns.boxplot(x=df4_num_mean[var], orient='h')
    plt.subplot(3, 1, 3)
    sns.boxplot(x=df5_num_median[var], orient='h')
```
