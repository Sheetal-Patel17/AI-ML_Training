### Types of missing data

| Type                                    | Meaning                                         | Example                                                             |
| --------------------------------------- | ----------------------------------------------- | ------------------------------------------------------------------- |
| **MCAR** — Missing Completely At Random | Missingness is unrelated to any data            | A sensor randomly fails                                             |
| **MAR** — Missing At Random             | Missingness depends on other observed variables | Income is missing more often for people in a particular age group   |
| **MNAR** — Missing Not At Random        | Missingness depends on the missing value itself | People with very high income are less likely to report their income |


**Common methods for filling missing values**

1. Mean imputation — numerical data
2. Median imputation — numerical data, especially when outliers exist
3. Mode imputation — categorical data
4. Forward fill / backward fill — time-series data
5. Constant value — e.g., "Unknown" for categorical data
6. KNN imputation — estimates values using similar observations
7. Iterative imputation — predicts missing values using other features





- Numerical → median/mean
- Categorical → mode/most frequent
- Time series → forward/backward fill
- More advanced → KNN or Iterative Imputer

## MCAR — Missing Completely At Random

```python
import numpy as np
import pandas as pd

np.random.seed(42)

df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40, 45, 50, 55, 60, 65],
    "Salary": [25000, 30000, 35000, 40000, 45000,
               50000, 55000, 60000, 65000, 70000]
})

# Randomly make 20% of Salary values missing
mask = np.random.rand(len(df)) < 0.2
df.loc[mask, "Salary"] = np.nan

print(df)
```

## MAR — Missing At Random

```python
import numpy as np
import pandas as pd

np.random.seed(42)

df = pd.DataFrame({
    "Age": [20, 22, 25, 30, 35, 40, 45, 50, 55, 60],
    "Salary": [25000, 28000, 32000, 40000, 45000,
               50000, 55000, 60000, 65000, 70000]
})

# Salary is more likely to be missing when Age < 30
probability = np.where(df["Age"] < 30, 0.7, 0.1)

mask = np.random.rand(len(df)) < probability
df.loc[mask, "Salary"] = np.nan

print(df)
```

## MNAR — Missing Not At Random

```python
import numpy as np
import pandas as pd

np.random.seed(42)

df = pd.DataFrame({
    "Age": [25, 30, 35, 40, 45, 50, 55, 60, 65, 70],
    "Salary": [25000, 30000, 35000, 45000, 50000,
               60000, 80000, 100000, 150000, 200000]
})

# High-salary people are more likely to have missing salary
probability = np.where(df["Salary"] > 70000, 0.8, 0.1)

mask = np.random.rand(len(df)) < probability
df.loc[mask, "Salary"] = np.nan

print(df)
```
